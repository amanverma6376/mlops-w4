```python
import mlflow
import pandas as pd
from iris_pipeline_mlflow import IrisMLflowPipeline
import mlflow_config
```

```python
tracking_uri, artifact_root = mlflow_config.setup_mlflow_gcs()
print(f"MLflow Tracking URI: {tracking_uri}")
print(f"Artifact Root: {artifact_root}")
```

```python
pipeline = IrisMLflowPipeline()

results = pipeline.run_experiment()

for model_type, result in results.items():
    print(f"\n{model_type.upper()} Results:")
    print(f"  Test Accuracy: {result['test_accuracy']:.4f}")
    print(f"  CV Score: {result['cv_score']:.4f}")
    print(f"  Run ID: {result['run_id']}")
    print(f"  Best Params: {result['best_params']}")
```

```python
experiment = mlflow.get_experiment_by_name("iris-hyperparameter-tuning")
runs = mlflow.search_runs(experiment_ids=[experiment.experiment_id])

top_runs = runs.sort_values('metrics.test_accuracy', ascending=False).head(5)
print("Top 5 Runs by Test Accuracy:")
print(top_runs[['run_id', 'metrics.test_accuracy', 'params.model_type']].to_string())
```
